# Select and Filter Data

**Topics Covered**
- Select Specific Columns
- Filter Rows
- Combine Multiple Conditions

---

**Select Columns**

**Single Column (Series)**

```python
column = df["column_name"]
```

**Multiple Columns (DataFrame)**

```python
subset = df[["col1", "col2", "col3"]]
```

---

**Filter Rows**

**Single Condition**

```python
filtered_rows = df[df["salary"] > 50000]
```

**Multiple Conditions (AND)**

```python
filtered_rows = df[(df["salary"] > 50000) & (df["salary"] < 80000)]
```

**Multiple Conditions (OR)**

```python
filtered_rows = df[(df["city"] == "Mumbai") | (df["city"] == "Pune")]
```

**NOT Condition**

```python
filtered_rows = df[~(df["city"] == "Mumbai")]
```

---

**Operators**
- `&` → AND
- `|` → OR
- `~` → NOT

````python
#single condition 
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha", "Rohit", "Anjali", "Vikram"],
    "Age": [22, 25, 24, 28, 23, 27, 26, 29, 21, 30],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000, 62000, 80000, 42000, 90000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good", "Excellent", "Average", "Excellent", "Good", "Excellent"]
}

df = pd.DataFrame(data)
highest_salary=df[df["Salary"]<50000]
print(highest_salary)
````

````python
#multiple  condition 
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha", "Rohit", "Anjali", "Vikram"],
    "Age": [22, 25, 24, 28, 23, 27, 26, 29, 21, 30],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000, 62000, 80000, 42000, 90000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good", "Excellent", "Average", "Excellent", "Good", "Excellent"]
}

df = pd.DataFrame(data)
highest_salary=df[(df["Salary"]>50000) & (df["Age"]>25) ]
print(highest_salary)
````
