# Information of Data (`info()`)

**Why use `info()`?**
- Understand the DataFrame structure.
- Check column names, data types, and missing values.

**`info()` displays:**
- Rows and columns
- Column names
- Data types
- Non-null values
- Memory usage

```python
import pandas  as pd
df=pd.read_json(r"..\Datasets\sample_Data.json")

print("display info of data")
print(df.info())
```
