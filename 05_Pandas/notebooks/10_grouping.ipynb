# Grouping (GroupBy)

**Definition:**  
Groups data based on one or more columns and performs calculations on each group.

| Topic | Syntax | Example |
|-------|--------|---------|
| Group by One Column | `df.groupby("Column")` | `df.groupby("Performance")` |
| Group + Sum | `df.groupby("Column").sum()` | `df.groupby("Performance")["Salary"].sum()` |
| Group + Mean | `df.groupby("Column").mean()` | `df.groupby("Performance")["Salary"].mean()` |
| Group + Count | `df.groupby("Column").count()` | `df.groupby("Performance")["Salary"].count()` |
| Group + Min | `df.groupby("Column").min()` | `df.groupby("Performance")["Salary"].min()` |
| Group + Max | `df.groupby("Column").max()` | `df.groupby("Performance")["Salary"].max()` |
| Group + Multiple Aggregations | `df.groupby("Column").agg([...])` | `df.groupby("Performance")["Salary"].agg(["sum","mean","min","max"])` |
| Group by Multiple Columns | `df.groupby(["Col1","Col2"])` | `df.groupby(["Performance","Age"])["Salary"].sum()` |

```python
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha", "Karan"],
    "Age": [22, 25, 24, 28, 23, 27],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good", "Excellent"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Group by Performance
print("Group by Performance:")
print(df.groupby("Performance"))
```

```python
# Group by Performance and Sum of Salary
print("Salary Sum by Performance:")
print(df.groupby("Performance")["Salary"].sum())
```

```python
# Group by Performance and Average Salary
print("Average Salary by Performance:")
print(df.groupby("Performance")["Salary"].mean())
```

```python
# Group by Performance and Count
print("Employee Count by Performance:")
print(df.groupby("Performance")["Salary"].count())
```

```python
# Group by Performance and Minimum Salary
print("Minimum Salary by Performance:")
print(df.groupby("Performance")["Salary"].min())
```

```python
# Multiple Aggregations
print("Multiple Aggregations:")
print(df.groupby("Performance")["Salary"].agg(["sum", "mean", "min", "max", "count"]))
```

```python
# Group by Multiple Columns
print("Group by Performance and Age:")
print(df.groupby(["Performance", "Age"])["Salary"].sum())
```
