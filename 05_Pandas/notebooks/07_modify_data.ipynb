# Modify Data

| Topic | Syntax | Example |
|-------|--------|---------|
| Add New Column | `df["Column_Name"] = values` | `df["Bonus"] = df["Salary"] * 0.10` |
| Modify Existing Column | `df["Column_Name"] = new_values` | `df["Salary"] = df["Salary"] + 5000` |
| Modify Specific Value | `df.loc[row_index, "Column_Name"] = value` | `df.loc[0, "Salary"] = 50000` |
| Modify Multiple Rows | `df.loc[condition, "Column_Name"] = value` | `df.loc[df["Age"] > 25, "Salary"] += 2000` |
| Rename Column | `df.rename(columns={"Old_Name":"New_Name"})` | `df.rename(columns={"Salary":"Income"})` |
| Insert Column | `df.insert(position, "Column_Name", values)` | `df.insert(1, "Department", dept)` |
| Delete Column | `df.drop(columns=["Column_Name"])` | `df.drop(columns=["Age"])` |
| Delete Row | `df.drop(index=row_index)` | `df.drop(index=0)` |
| Replace Values | `df["Column_Name"].replace(old, new)` | `df["Performance"].replace("Good","Average")` |
| Change Data Type | `df["Column_Name"].astype(data_type)` | `df["Age"] = df["Age"].astype(int)` |
| Apply Function | `df["New_Column"] = df["Column_Name"].apply(function)` | `df["Tax"] = df["Salary"].apply(lambda x: x*0.05)` |
| Map Values | `df["New_Column"] = df["Column_Name"].map(dictionary)` | `df["Rating"] = df["Performance"].map(rating)` |

```python
#add column
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha", "Rohit", "Anjali", "Vikram"],
    "Age": [22, 25, 24, 28, 23, 27, 26, 29, 21, 30],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000, 62000, 80000, 42000, 90000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good", "Excellent", "Average", "Excellent", "Good", "Excellent"]
}

df = pd.DataFrame(data)
print(df)
df["bonus"]=df["Salary"]*0.10

print("\nDataFrame after adding Bonus column:")
print(df)
```
