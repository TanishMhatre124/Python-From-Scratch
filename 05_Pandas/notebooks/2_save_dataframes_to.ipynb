```python
import pandas as pd
```

```python
#save data in pandas in csv
data={
    "name":["tan","jash","tanish"],
    "age":["10","20","30"],
    "city":["mumbai","pune","delhi"]
}

df= pd.DataFrame(data)
print(df)

df.to_csv("../outputs/output.csv", index=False)
#save data in pandas in json

data={
    "name":["tan","jash","tanish"],
    "age":["10","20","30"],
    "city":["mumbai","pune","delhi"]
}

df= pd.DataFrame(data)
print(df)

df.to_json("../outputs/output.json", index=False)
```

```python
#save data in pandas in excel

data={
    "name":["tan","jash","tanish"],
    "age":["10","20","30"],
    "city":["mumbai","pune","delhi"]
}

df= pd.DataFrame(data)
print(df)

df.to_excel("../outputs/output.xlsx", index=False)
```
