# Merging and Joining

**Definition:**  
Combines two or more DataFrames based on a common column or index.

| Topic | Syntax | Example |
|-------|--------|---------|
| Inner Merge | `pd.merge(df1, df2, on="Column")` | `pd.merge(df1, df2, on="ID")` |
| Left Merge | `pd.merge(df1, df2, on="Column", how="left")` | `pd.merge(df1, df2, on="ID", how="left")` |
| Right Merge | `pd.merge(df1, df2, on="Column", how="right")` | `pd.merge(df1, df2, on="ID", how="right")` |
| Outer Merge | `pd.merge(df1, df2, on="Column", how="outer")` | `pd.merge(df1, df2, on="ID", how="outer")` |
| Merge on Multiple Columns | `pd.merge(df1, df2, on=["Col1","Col2"])` | `pd.merge(df1, df2, on=["ID","Name"])` |
| Join by Index | `df1.join(df2)` | `df1.join(df2)` |
| Left Join | `df1.join(df2, how="left")` | `df1.join(df2, how="left")` |
| Right Join | `df1.join(df2, how="right")` | `df1.join(df2, how="right")` |
| Outer Join | `df1.join(df2, how="outer")` | `df1.join(df2, how="outer")` |

```python
import pandas as pd

employees = {
    "ID": [101, 102, 103, 104],
    "Name": ["Tanish", "Rahul", "Priya", "Amit"]
}

salary = {
    "ID": [101, 102, 104, 105],
    "Salary": [45000, 60000, 75000, 50000]
}

df1 = pd.DataFrame(employees)
df2 = pd.DataFrame(salary)

print(df1)
print(df2)
```

```python
# Inner Merge
# Combines only the rows that have matching values in both DataFrames.
print("Inner Merge:")
print(pd.merge(df1, df2, on="ID"))
```

```python
# Left Merge
# Keeps all rows from the left DataFrame and adds matching rows from the right DataFrame.
print("Left Merge:")
print(pd.merge(df1, df2, on="ID", how="left"))
```

```python
# Right Merge
# Keeps all rows from the right DataFrame and adds matching rows from the left DataFrame.
print("Right Merge:")
print(pd.merge(df1, df2, on="ID", how="right"))
```

```python
# Outer Merge
# Keeps all rows from both DataFrames and fills missing values with NaN.
print("Outer Merge:")
print(pd.merge(df1, df2, on="ID", how="outer"))
```

```python
# Merge on Multiple Columns
# Combines DataFrames using more than one common column.
print("Merge on Multiple Columns:")
print(pd.merge(df1, df2, on=["ID"]))
```

**Joins**

```python
import pandas as pd

# Employee Data
employees = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit"]
}

# Salary Data
salary = {
    "Salary": [45000, 60000, 75000]
}

df1 = pd.DataFrame(employees, index=[0, 1, 2, 3])
df2 = pd.DataFrame(salary, index=[1, 2, 4])

print("Employee DataFrame")
print(df1)

print("\nSalary DataFrame")
print(df2)
```

```python
# Join by Index
# Joins two DataFrames using their row indexes.
print("Join by Index:")
print(df1.join(df2, rsuffix="_salary"))
```

```python
# Left Join
# Keeps all indexes from the left DataFrame.
print(df1.join(df2, how="left"))
```

```python
# Right Join
# Keeps all indexes from the right DataFrame.
print(df1.join(df2, how="right"))
```

```python
# Inner Join
# Keeps only the common indexes in both DataFrames.
print(df1.join(df2, how="inner"))
```

```python
# Outer Join
# Keeps all indexes from both DataFrames.
print(df1.join(df2, how="outer"))
```
