**describe data**

```python
import pandas as pd

data = {
    "name": ["tan", "jash", "tanish", "parth"],
    "age": [10, 20, 30, 10],  
    "city": ["mumbai", "pune", "delhi", "thane"]
}

df = pd.DataFrame(data)
print(df)

print("===================================")
print(df.dtypes)
print("===================================")
print("descriptive stastics ")
print(df.describe(include="all"))
```

```python
import pandas as pd

data = {
    "Name": ["Tanish", "Jash", "Parth", "Riya"],
    "Age": [22, 21, 23, 20],
    "City": ["Mumbai", "Pune", "Thane", "Delhi"],
    "Salary": [50000, 60000, 55000, 70000]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)
```
