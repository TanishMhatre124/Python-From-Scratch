# Handling Missing Data
Interpolation fills missing numeric values by estimating them from the values before and after the missing value.

| Topic | Syntax | Example |
|-------|--------|---------|
| Check Missing Values | `df.isnull()` | `df.isnull()` |
| Count Missing Values | `df.isnull().sum()` | `df.isnull().sum()` |
| Check Non-Missing Values | `df.notnull()` | `df.notnull()` |
| Remove Rows with Missing Values | `df.dropna()` | `df.dropna()` |
| Remove Columns with Missing Values | `df.dropna(axis=1)` | `df.dropna(axis=1)` |
| Fill Missing Values | `df.fillna(value)` | `df.fillna(0)` |
| Fill with Mean | `df["Column"].fillna(df["Column"].mean())` | `df["Age"].fillna(df["Age"].mean())` |
| Fill with Median | `df["Column"].fillna(df["Column"].median())` | `df["Salary"].fillna(df["Salary"].median())` |
| Fill with Mode | `df["Column"].fillna(df["Column"].mode()[0])` | `df["City"].fillna(df["City"].mode()[0])` |
| Forward Fill | `df.ffill()` | `df.ffill()` |
| Backward Fill | `df.bfill()` | `df.bfill()` |
| Replace Missing Values | `df.replace(old, new)` | `df.replace(np.nan, 0)` |

```python
#null and sum of missisng value
import pandas as pd

data = {
    "Name": ["Tanish", None, "Priya", "Amit", "Sneha", "Karan", "Neha", "Rohit", "Anjali", "Vikram"],
    "Age": [22, 25, 24, None, 23, 27, 26, 29, 21, 30],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000, None, 80000, 42000, 90000],
    "Performance": ["Excellent", "Good", "Average", None, "Good", "Excellent", "Average", "Excellent", "Good", "Excellent"]
}


df= pd.DataFrame(data)
print(df.isnull())
print(df.isnull().sum())

```

```python
#fill missisng value with data 
import pandas as pd

data = {
    "Name": ["Tanish", None, "Priya", "Amit", "Sneha", "Karan", "Neha", "Rohit", "Anjali", "Vikram"],
    "Age": [22, 25, 24, None, 23, 27, 26, 29, 21, 30],
    "Salary": [45000, 60000, 55000, 75000, 50000, 68000, None, 80000, 42000, 90000],
    "Performance": ["Excellent", "Good", "Average", None, "Good", "Excellent", "Average", "Excellent", "Good", "Excellent"]
}


df= pd.DataFrame(data)
print(df)


df["Name"] = df["Name"].fillna("Unknown")
df["Age"] = df["Age"].fillna(0)
df["Salary"] = df["Salary"].fillna(df["Salary"].mean(),inplace=True)
print("\n",df)
```

# Interpolation

| Topic | Syntax | Example |
|-------|--------|---------|
| Linear Interpolation | `df.interpolate()` | `df.interpolate()` |
| Update Original Data | `df.interpolate(inplace=True)` | `df.interpolate(inplace=True)` |
| Interpolate Specific Column | `df["Column"].interpolate()` | `df["Salary"].interpolate()` |
| Polynomial Interpolation | `df.interpolate(method="polynomial", order=2)` | `df.interpolate(method="polynomial", order=2)` |
| Time Interpolation | `df.interpolate(method="time")` | `df.interpolate(method="time")` |

```python
import pandas as pd
import numpy as np

data = {
    "Age": [22, np.nan, 24, np.nan, 28],
    "Salary": [45000, np.nan, 55000, np.nan, 75000],
    "Experience": [1, np.nan, 3, np.nan, 5]
}

df = pd.DataFrame(data)

print("Before Interpolation:")
print(df)
df = df.interpolate(method="linear")

print("\nAfter Interpolation:")
print(df)
```
