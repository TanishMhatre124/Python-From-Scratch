# Concatenate

**Definition:**  
Concatenates (combines) two or more DataFrames either row-wise or column-wise.

| Topic | Syntax | Example |
|-------|--------|---------|
| Concatenate Rows | `pd.concat([df1, df2])` | `pd.concat([df1, df2])` |
| Concatenate Columns | `pd.concat([df1, df2], axis=1)` | `pd.concat([df1, df2], axis=1)` |
| Ignore Index | `pd.concat([df1, df2], ignore_index=True)` | `pd.concat([df1, df2], ignore_index=True)` |
| Outer Concatenation (Default) | `pd.concat([df1, df2], join="outer")` | `pd.concat([df1, df2], join="outer")` |
| Inner Concatenation | `pd.concat([df1, df2], join="inner")` | `pd.concat([df1, df2], join="inner")` |

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [101, 102],
    "Name": ["Tanish", "Rahul"]
})

df2 = pd.DataFrame({
    "ID": [103, 104],
    "Name": ["Priya", "Amit"]
})

print("DataFrame 1")
print(df1)

print("\nDataFrame 2")
print(df2)
```

```python
# Ignore Index
# Combines DataFrames and creates a new continuous index.
print("Ignore Index:")
print(pd.concat([df1, df2], ignore_index=True))
```

```python
df3 = pd.DataFrame({
    "Salary": [45000, 60000]
})
```

```python
# Concatenate Columns
# Adds DataFrames side by side.
print("Concatenate Columns:")
print(pd.concat([df1, df3], axis=1))
```

```python
df4 = pd.DataFrame({
    "Salary": [45000]
})
```

```python
# Outer Concatenation
# Keeps all rows and fills missing values with NaN.
print("Outer Concatenation:")
print(pd.concat([df1, df4], axis=1, join="outer"))
```

```python
# Inner Concatenation
# Keeps only the common indexes.
print("Inner Concatenation:")
print(pd.concat([df1, df4], axis=1, join="inner"))
```
