# Handling Missing Data

| Topic | Syntax | Example |
|-------|--------|---------|
| Check Missing Values | `df.isnull()` | `df.isnull()` |
| Count Missing Values | `df.isnull().sum()` | `df.isnull().sum()` |
| Check Non-Missing Values | `df.notnull()` | `df.notnull()` |
| Remove Rows with Missing Values | `df.dropna()` | `df.dropna()` |
| Remove Columns with Missing Values | `df.dropna(axis=1)` | `df.dropna(axis=1)` |
| Fill Missing Values | `df.fillna(value)` | `df.fillna(0)` |
| Fill with Mean | `df["Column"].fillna(df["Column"].mean())` | `df["Age"].fillna(df["Age"].mean())` |
| Fill with Median | `df["Column"].fillna(df["Column"].median())` | `df["Salary"].fillna(df["Salary"].median())` |
| Fill with Mode | `df["Column"].fillna(df["Column"].mode()[0])` | `df["City"].fillna(df["City"].mode()[0])` |
| Forward Fill | `df.ffill()` | `df.ffill()` |
| Backward Fill | `df.bfill()` | `df.bfill()` |
| Replace Missing Values | `df.replace(old, new)` | `df.replace(np.nan, 0)` |
