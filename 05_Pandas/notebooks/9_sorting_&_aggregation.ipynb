# Sorting

**Definition:**  
Sorts data in ascending or descending order.

| Topic | Syntax | Example |
|-------|--------|---------|
| Sort by One Column | `df.sort_values("Column")` | `df.sort_values("Salary")` |
| Descending Order | `df.sort_values("Column", ascending=False)` | `df.sort_values("Salary", ascending=False)` |
| Sort by Multiple Columns | `df.sort_values(["Col1","Col2"])` | `df.sort_values(["Age","Salary"])` |
| Sort by Multiple Columns (Different Order) | `df.sort_values(["Age","Salary"], ascending=[True, False])` | `df.sort_values(["Age","Salary"], ascending=[True, False])` |
| Sort by Index | `df.sort_index()` | `df.sort_index()` |
| Update Original Data | `df.sort_values("Column", inplace=True)` | `df.sort_values("Salary", inplace=True)` |

```python
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha"],
    "Age": [22, 25, 24, 28, 23],
    "Salary": [45000, 60000, 55000, 75000, 50000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Sort by Salary (Ascending)
print("Sort by Salary (Ascending):")
print(df.sort_values("Salary"))
```

```python
# Sort by Salary (Descending)
print("Sort by Salary (Descending):")
print(df.sort_values("Salary", ascending=False))
```

```python
# Sort by Age (Ascending) and Salary (Descending)
print("Sort by Age (Ascending) and Salary (Descending):")
print(df.sort_values(["Age", "Salary"], ascending=[True, False]))
```

```python
# Sort by Index
print("Sort by Index:")
print(df.sort_index())
```

```python
# Sort Original DataFrame
df.sort_values("Salary", inplace=True)
print("Original DataFrame Sorted by Salary:")
print(df)
```

# Aggregation

**Definition:**  
Performs calculations on one or more columns to summarize data.

| Topic | Syntax | Example |
|-------|--------|---------|
| Sum | `df["Column"].sum()` | `df["Salary"].sum()` |
| Mean (Average) | `df["Column"].mean()` | `df["Salary"].mean()` |
| Median | `df["Column"].median()` | `df["Salary"].median()` |
| Minimum | `df["Column"].min()` | `df["Salary"].min()` |
| Maximum | `df["Column"].max()` | `df["Salary"].max()` |
| Count Values | `df["Column"].count()` | `df["Salary"].count()` |
| Standard Deviation | `df["Column"].std()` | `df["Salary"].std()` |
| Variance | `df["Column"].var()` | `df["Salary"].var()` |
| Multiple Aggregations | `df.agg(["sum","mean","min","max"])` | `df.agg(["sum","mean","min","max"])` |
| Aggregate Specific Column | `df["Column"].agg(["sum","mean"])` | `df["Salary"].agg(["sum","mean"])` |

```python
import pandas as pd

data = {
    "Name": ["Tanish", "Rahul", "Priya", "Amit", "Sneha"],
    "Age": [22, 25, 24, 28, 23],
    "Salary": [45000, 60000, 55000, 75000, 50000],
    "Performance": ["Excellent", "Good", "Average", "Excellent", "Good"]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Sum of Salary
print("Total Salary:")
print(df["Salary"].sum())
```

```python
# Average Salary
print("Average Salary:")
print(df["Salary"].mean())
```

```python
# Median Salary
print("Median Salary:")
print(df["Salary"].median())
```

```python
# Minimum Salary
print("Minimum Salary:")
print(df["Salary"].min())
```

```python
# Maximum Salary
print("Maximum Salary:")
print(df["Salary"].max())
```

```python
# Count of Salary Values
print("Count of Salary:")
print(df["Salary"].count())
```

```python
# Standard Deviation
print("Standard Deviation:")
print(df["Salary"].std())
```

```python
# Variance
print("Variance:")
print(df["Salary"].var())
```

```python
# Multiple Aggregations on Salary
print("Multiple Aggregations:")
print(df["Salary"].agg(["sum", "mean", "min", "max"]))
```

```python
# Aggregate All Numeric Columns
print("Aggregate All Numeric Columns:")
df.select_dtypes(include="number").agg(["sum", "mean", "min", "max"])
```
